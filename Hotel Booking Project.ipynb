# **Problem Statement -**

The hotel booking industry faces various challenges in improving operational efficiency, customer satisfaction, and maximizing revenue. The analysis of booking data can uncover valuable insights to address these challenges and optimize business outcomes. The primary objectives of this project are:

1. **High Cancellation Rates**  
   A significant percentage of hotel bookings are canceled, leading to lost revenue. This project aims to identify the key factors influencing booking cancellations and propose actionable strategies to reduce cancellation rates, ultimately boosting hotel profitability.

2. **Predicting Customer Behavior**  
   Understanding customer behavior is critical for personalized marketing and optimizing service offerings. This project will leverage historical booking data to predict customer booking behavior, including likelihood of cancellation, booking patterns, and customer preferences, to enhance marketing strategies and customer engagement.

3. **Revenue Optimization**  
   The hotel is unable to effectively optimize room pricing and availability, resulting in lost revenue opportunities. This analysis aims to examine booking patterns, customer segments, and market conditions to provide data-driven recommendations for dynamic pricing strategies and better inventory management to maximize revenue.

4. **Impact of Seasonal Trends**  
   Seasonal variations in booking demand present challenges in resource allocation and revenue generation. This project will explore seasonal trends in booking data, providing insights into how the hotel can adjust marketing strategies, pricing, and resource management to capitalize on peak seasons and optimize bookings during off-seasons.

5. **Customer Satisfaction & Service Quality**  
   Measuring and improving customer satisfaction based on booking data alone can be challenging. This analysis aims to integrate booking data with customer feedback to identify areas of improvement in the booking process, service quality, and overall guest experience to increase customer retention and loyalty.

6. **Performance of Booking Channels**  
   The hotel struggles to determine which booking channels (e.g., direct bookings, third-party agencies, or travel websites) bring in the most profitable and loyal customers. This project will analyze data across different booking channels to assess their performance and offer recommendations on optimizing the booking strategy by focusing on the most effective channels.

7. **Influence of Booking Features on Cancellations**  
   Specific booking features, such as refundable vs. non-refundable rates, number of guests, and length of stay, may influence the likelihood of cancellations. The project will explore the impact of these features on cancellation rates and suggest improvements to booking policies and product offerings to minimize cancellations.

# **Variable Description -:**


1. **hotel :** Name of the hotel (Resort Hotel or City Hotel)

2. **is_canceled :** If the booking was canceled (1) or not (0)

3. **lead_time:** Number of days before the actual arrival of the guests

4. **arrival_date_year :** Year of arrival date

5. **arrival_date_month :** Month of arrival date

6. **arrival_date_week_number :** Week number of year for arrival date

7. **arrival_date_day_of_month :** Day of arrival date

8. **stays_in_weekend_nights :** Number of weekend nights (Saturday or Sunday) spent at the hotel by the guests.

9. **stays_in_week_nights :** Number of weeknights (Monday to Friday) spent at the hotel by the guests.

10. **adults :** Number of adults among guests

11. **children :** Number of children among guests

12. **babies :** Number of babies among guests

13. **meal :** Type of meal booked
- **BB**: Bed and Breakfast (includes breakfast only)  
- **FB**: Full Board (includes all meals: breakfast, lunch, and dinner)  
- **HB**: Half Board (includes breakfast and either lunch or dinner)  
- **SC**: Self-Catering (no meals included)  
- **Undefined**: No specific board type or undefined data  

14. **country :** Country of guests

15. **market_segment :** Designation of market segment 

- **Direct**: Bookings made directly with the hotel.  
- **Corporate**: Bookings made by corporate clients or for business purposes.  
- **Online TA**: Bookings made via online travel agents.  
- **Offline TA/TO**: Bookings made via offline travel agents or tour operators.  
- **Complementary**: Bookings offered as complimentary (e.g., free stays for promotions or rewards).  
- **Groups**: Bookings made for large groups (e.g., events or conferences).  
- **Undefined**: Booking channel not specified or unknown.  
- **Aviation**: Bookings made as part of aviation-related services (e.g., airline crew stays).  

16. **distribution_channel :** Name of booking distribution channel

- **Direct**: Bookings made directly with the hotel.  
- **Corporate**: Bookings made by corporate clients or for business purposes.  
- **TA/TO**: Bookings made via travel agents or tour operators.  
- **Undefined**: Market segment not specified or unknown.  
- **GDS**: Bookings made through the Global Distribution System (used by travel agents and corporate clients to book hotels).  


17. **is_repeated_guest :** If the booking was from a repeated guest (1) or not (0)

18. **previous_cancellations :** Number of previous bookings that were cancelled by the customer prior to the current booking

19. **previous_bookings_not_canceled :** Number of previous bookings not cancelled by the customer prior to the current booking

20. **reserved_room_type :** Code of room type reserved

21. **assigned_room_type :** Code of room type assigned

22. **booking_changes :** Number of changes/amendments made to the booking

23. **deposit_type :** Type of the deposit made by the guest

24. **agent :** ID of travel agent who made the booking

25. **company :** ID of the company that made the booking

26. **days_in_waiting_list :** Number of days the booking was in the waiting list

27. **customer_type :** Type of customer

- **Transient**: Individual bookings that are not part of a group or contract.  
- **Contract**: Bookings associated with a contract, typically negotiated with a company or organization.  
- **Transient-Party**: Bookings made by individuals traveling as part of a larger group, but not classified as a "Group" booking.  
- **Group**: Bookings made for a large group, such as for events, conferences, or tours.  


28. **adr :** Average Daily Rate, as defined by dividing the sum of all lodging transactions by the total number of staying nights

29. **required_car_parking_spaces :** Number of car parking spaces required by the customer

30. **total_of_special_requests :** Number of special requests made by the customer

31. **reservation_status :** Reservation status (Canceled, Check-Out or No-Show)

32. **reservation_status_date :** Date at which the last reservation status was updated

## 3. ***Data Wrangling***

```python
#Droping The Company And Agent Because There Is Large Amount Of Data Are Missing I Choose Droping These Column will be a Good Choice
hotel.drop(columns=['company','agent'],inplace=True)
```

```python
# Replacing null values of children columns with 0 and country column with 'others'
# In children column there are 4 Null values and in Company column there are 488 Null values

hotel['children'] == hotel['children'].fillna(0)
hotel['country'] = hotel['country'].fillna('others')
```

```python
# Checking number of duplicate values in the dataset
print('Total Count Of Duplicated Rows : ',hotel.duplicated().sum())
```

```python
# Dropping the duplicate rows from dataset
hotel = hotel.drop_duplicates()
```

```python
# Checking how many rows in this dataset whose combining values of adults, babies and children columns is 0
hotel[['adults','babies','children']][hotel['adults'] + hotel['babies'] + hotel['children'] == 0].shape[0]
```

```python
# Dropping the rows where combining values of adults, babies and children columns is 0, because it simply means no bookings were made
hotel.drop(hotel[['adults','babies','children']][hotel['adults'] + hotel['babies'] + hotel['children'] == 0].index,inplace=True)
```

### **Converting columns to appropriate datatypes**

```python
hotel['reservation_status_date'] = pd.to_datetime(hotel['reservation_status_date'])
```

### **Adding important columns as per requirement**

```python
hotel['Total_Stays'] = hotel['stays_in_week_nights'] + hotel['stays_in_weekend_nights']

hotel['Total_Member'] = hotel['adults'] + hotel['babies'] + hotel['children']
```
